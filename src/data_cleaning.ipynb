```python
from pypdf import PdfReader
```

```python
reader = PdfReader("../data/thinkpython2.pdf")
print("Pages:", len(reader.pages))
```

```python
text = ""

for page in reader.pages:
    page_text = page.extract_text()
    if page_text:
        text += page_text + "\n"

print("Extraction complete")
```

```python
print(text[:3000])
```

```python
cleaned_text = text.replace("\n", " ")
cleaned_text = " ".join(cleaned_text.split())

print(cleaned_text[:1000])
```

```python
with open("../data/cleaned_thinkpython2.txt", "w", encoding="utf-8") as f:
    f.write(cleaned_text)

print("Saved successfully!")
```

```python
import pandas as pd
import os

# Raw and cleaned file pairs
datasets = {
    'academic_ai_excel': ('academic_ai_raw_data.xlsx', 'cleaned_academic_ai_excel.csv'),
    'ai_anxiety': ('ai_dependency_career_anxiety_students.csv', 'cleaned_ai_anxiety.csv'),
    'coursea_data': ('coursea_data.csv', 'cleaned_coursea_data.csv'),
    'essay_assistance': ('essay_assistance_raw.csv', 'cleaned_essay_assistance.csv'),
    'python_faq': ('python_faq_dataset.csv', 'cleaned_python_faq.csv'),
    'qa_interactions': ('qa_interactions_raw.csv', 'cleaned_qa_interactions.csv'),
    'student_por': ('student-por.csv', 'cleaned_student_por.csv'),
    'studentAssessment': ('studentAssessment.csv', 'cleaned_studentAssessment.csv'),
    'students_performance': ('StudentsPerformance.csv', 'cleaned_students_performance.csv'),
    'study_schedules': ('study_schedules_raw.csv', 'cleaned_study_schedules.csv'),
}

raw_folder = '../data/raw/'
cleaned_folder = '../data/cleaned/'

print(f"{'Dataset':<25} {'Raw Rows':<12} {'Cleaned Rows':<14} {'Rows Removed':<12}")
print("-" * 65)

for name, (raw_file, clean_file) in datasets.items():
    try:
        raw_path = raw_folder + raw_file
        clean_path = cleaned_folder + clean_file
        
        if raw_file.endswith('.xlsx'):
            raw_df = pd.read_excel(raw_path)
        else:
            raw_df = pd.read_csv(raw_path)
            
        clean_df = pd.read_csv(clean_path)
        
        raw_rows = len(raw_df)
        clean_rows = len(clean_df)
        removed = raw_rows - clean_rows
        
        print(f"{name:<25} {raw_rows:<12} {clean_rows:<14} {removed:<12}")
    except Exception as e:
        print(f"{name:<25} Error: {e}")
```
